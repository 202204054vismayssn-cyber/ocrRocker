```python
# ==============================================================================
# SCRIPT 1: TEST PyMuPDF STANDALONE (Digital Text Extraction)
# ==============================================================================
import fitz  # PyMuPDF
import time

# Path to your invoice PDF (change filename if testing your uploaded invoice)
PDF_PATH = '/content/india_sample.pdf'

start_time = time.time()
doc = fitz.open(PDF_PATH)
total_pages = len(doc)
full_text = ''

for page_num in range(total_pages):
    page = doc[page_num]
    text = page.get_text()
    if text.strip():
        full_text += f'\n--- PAGE {page_num + 1} ---\n' + text.strip()

elapsed = time.time() - start_time

print(f'=== PyMuPDF RESULTS ===')
print(f'File: {PDF_PATH}')
print(f'Total Pages: {total_pages}')
print(f'Time Taken: {elapsed:.4f} seconds')
print(f'Total Characters Found: {len(full_text)}')

if len(full_text.strip()) > 50:
    print('\n✅ SUCCESS: Document is a DIGITAL PDF! Clean text extracted instantly:')
    print('----------------------------------------------------------------------')
    print(full_text[:600] + ('\n... [truncated]' if len(full_text) > 600 else ''))
else:
    print('\n⚠️ WARNING: Document has NO digital text (it is a scanned image/photo).')
    print('This document requires OCR (Script 2).')
```

```python
# ==============================================================================
# SCRIPT 2: TEST PaddleOCR STANDALONE (OCR Engine on GPU)
# ==============================================================================
import fitz
import numpy as np
from PIL import Image
import io, time

PDF_PATH = '/content/india_sample.pdf'

# 1. Render first page to image using PyMuPDF
doc = fitz.open(PDF_PATH)
page = doc[0]
pix = page.get_pixmap(dpi=150)
img = Image.open(io.BytesIO(pix.tobytes('png'))).convert('RGB')
img_np = np.asarray(img)

print('Page rendered to image. Initializing PaddleOCR on GPU...')

# 2. Initialize PaddleOCR
from paddleocr import PaddleOCR
ocr = PaddleOCR(use_textline_orientation=False, lang='en', device='gpu')

start_time = time.time()

# 3. Run OCR on image (use predict instead of deprecated ocr with cls=False)
if hasattr(ocr, 'predict'):
    results = ocr.predict(img_np)
else:
    results = ocr.ocr(img_np)

elapsed = time.time() - start_time

# 4. Extract lines (handles modern .json format as well as legacy)
extracted_lines = []
for item in results:
    data = item.json.get('res', {}) if hasattr(item, 'json') else (item if isinstance(item, dict) else {})
    texts = data.get('rec_text', []) or data.get('rec_texts', [])
    if texts:
        for t in texts:
            if str(t).strip():
                extracted_lines.append(str(t).strip())
    elif isinstance(item, (list, tuple)):
        for line in item:
            if isinstance(line, (list, tuple)) and len(line) >= 2 and isinstance(line[1], (list, tuple)):
                extracted_lines.append(str(line[1][0]).strip())

print(f'\n=== PaddleOCR RESULTS ===')
print(f'Time Taken: {elapsed:.2f} seconds')
print(f'Total Lines Detected: {len(extracted_lines)}')

if extracted_lines:
    print('\n✅ SUCCESS: PaddleOCR successfully scanned the document on GPU!')
    print('----------------------------------------------------------------------')
    print('\n'.join(extracted_lines[:15]))
    if len(extracted_lines) > 15:
        print(f'... [{len(extracted_lines) - 15} more lines]')
else:
    print('\n❌ ERROR: PaddleOCR detected no text.')
```

```python
# ==============================================================================
# SCRIPT 3: TEST OLLAMA & GEMMA 2 STANDALONE (JSON Extraction)
# ==============================================================================
import requests, json, time, os

OLLAMA_URL = 'http://127.0.0.1:11434/api/chat'
MODEL = 'gemma2:2b'

# Check if text was already extracted by Script 1 or Script 2 in this session
if 'full_text' in globals() and full_text.strip():
    sample_text = full_text.strip()
    print("ℹ Using extracted text from Script 1 (PyMuPDF Digital Extraction)")
elif 'extracted_lines' in globals() and extracted_lines:
    sample_text = '\n'.join(extracted_lines)
    print("ℹ Using extracted text from Script 2 (PaddleOCR GPU Extraction)")
else:
    sample_text = """
TAX INVOICE
Union Bank of India
GSTIN: 27AAACU0564G1ZH
Invoice No : 8039530126000001
Date : 31-01-2026
Details of Recipient: BAJAJ GENERAL INSURANCE LIMITED
Address: 1st Floor, GE Plaza, Airport Road, Pune-411006
GSTIN: 27AABCB5730G1ZX
Item: INCOME A/C COMMISSION FROM BAJAJ ALLIANZ LTD. | HSN/SAC: 997161 | Taxable: 1048413.91
CGST: 94356.89
SGST/UTGST: 94356.89
Grand Total: 1237127.69
"""
    print("ℹ Using default sample invoice text")

# Target JSON Schema for India invoices
schema = {
    'type': 'object',
    'properties': {
        'vendor_name': {'type': 'string'},
        'vendor_gst': {'type': 'string'},
        'customer_name': {'type': 'string'},
        'customer_gst': {'type': 'string'},
        'customer_pan': {'type': 'string'},
        'invoice_number': {'type': 'string'},
        'invoice_date': {'type': 'string'},
        'taxable_amount': {'type': 'number'},
        'cgst_amount': {'type': 'number'},
        'sgst_amount': {'type': 'number'},
        'invoice_amount': {'type': 'number'},
        'hsn_sac_code': {'type': 'string'}
    },
    'required': ['vendor_name', 'invoice_number', 'invoice_amount']
}

prompt = f"""You are an expert invoice data extractor for India invoices.
Extract the structured data from the following invoice text.
Invoice Text:
{sample_text}
"""

payload = {
    'model': MODEL,
    'messages': [{'role': 'user', 'content': prompt}],
    'format': schema,
    'stream': False,
    'options': {'temperature': 0.0}
}

# 1. Health check
try:
    requests.get('http://127.0.0.1:11434/', timeout=5)
    print("✅ Ollama service is active at 127.0.0.1:11434")
except Exception as e:
    print(f"❌ Cannot reach Ollama: {e}. Make sure to run Cell 1 first!")

print(f"Sending invoice text to {MODEL} in Ollama...")
start_time = time.time()

# Timeout set to 180s to prevent ReadTimeout during cold-start or heavy grammar compilation
try:
    resp = requests.post(OLLAMA_URL, json=payload, timeout=180)
    elapsed = time.time() - start_time

    print(f'\n=== OLLAMA / GEMMA 2 RESULTS ===')
    print(f'HTTP Status: {resp.status_code}')
    print(f'Inference Time: {elapsed:.2f} seconds')

    if resp.status_code == 200:
        content = resp.json().get('message', {}).get('content', '')
        print('\n✅ SUCCESS: Gemma 2 returned structured JSON:')
        print('----------------------------------------------------------------------')
        try:
            parsed = json.loads(content)
            print(json.dumps(parsed, indent=2))
        except Exception:
            print(content)
    else:
        print(f'\n❌ ERROR: Ollama returned {resp.status_code}: {resp.text}')
except requests.exceptions.Timeout:
    print('\n❌ TIMEOUT: Request timed out after 180 seconds.')
    print('Check GPU allocation using `!ollama ps` or check `/content/ollama.log`.')
except Exception as e:
    print(f'\n❌ ERROR: {e}')

# Diagnostic: Check model allocation
print('\n=== Ollama GPU / Processor Status ===')
os.system("ollama ps")
```

```python
# ==============================================================================
# SCRIPT 4: TEST BATCH PROCESSING (Prepare -> Submit -> Status -> Result)
# ==============================================================================
import requests, json, time

BASE_URL = 'http://127.0.0.1:8000'
HEADERS = {'SAVVY_API_KEY': 'kBLaCcQmnuk7yTALLOtcC3iM77Xt5q8H'}

print("=======================================================")
print("STEP 1: PREPARE INVOICE OCR (POST /batch/prepare)")
print("=======================================================")
with open('/content/india_sample.pdf', 'rb') as f:
    files = {'file': ('india_sample.pdf', f, 'application/pdf')}
    prep_res = requests.post(f'{BASE_URL}/api/v1/india/invoices/batch/prepare', headers=HEADERS, files=files, timeout=60)

prep_data = prep_res.json()
print("Prepare Status:", prep_res.status_code)
print(f"Extracted Characters: {prep_data.get('character_count')}")
ocr_text = prep_data.get('character_stored', '')

print("\n=======================================================")
print("STEP 2: SUBMIT BATCH (POST /batch/submit)")
print("=======================================================")
# Submitting a batch of 3 invoices for parallel GPU extraction
batch_payload = {
    "items": [
        {"custom_id": "inv_001", "character_stored": ocr_text},
        {"custom_id": "inv_002", "character_stored": ocr_text},
        {"custom_id": "inv_003", "character_stored": ocr_text}
    ]
}

submit_res = requests.post(f'{BASE_URL}/api/v1/india/invoices/batch/submit', headers=HEADERS, json=batch_payload, timeout=30)
submit_data = submit_res.json()
print("Submit Status:", submit_res.status_code)
batch_job_name = submit_data.get('batch_job_name')
print("Batch Job Name:", batch_job_name)

print("\n=======================================================")
print("STEP 3: POLL BATCH STATUS (GET /batch/status/{batchJobName})")
print("=======================================================")
for attempt in range(1, 40):
    time.sleep(3)
    status_res = requests.get(f'{BASE_URL}/api/v1/india/invoices/batch/status/{batch_job_name}', headers=HEADERS, timeout=15)
    status_data = status_res.json()
    state = status_data.get('state')
    completed = status_data.get('completed_items')
    total = status_data.get('total_items')
    print(f"[{attempt * 3}s] State: {state} | Progress: {completed}/{total}")
    if state in ("COMPLETED", "FAILED"):
        break

print("\n=======================================================")
print("STEP 4: RETRIEVE BATCH RESULTS (GET /batch/result/{batchJobName})")
print("=======================================================")
result_res = requests.get(f'{BASE_URL}/api/v1/india/invoices/batch/result/{batch_job_name}', headers=HEADERS, timeout=30)
result_data = result_res.json()
print("Result Status:", result_res.status_code)
print(f"Total Completed: {result_data.get('completed_items')}/{result_data.get('total_items')}")

for item in result_data.get('results', []):
    print(f"\n--- Result for Item: {item.get('custom_id')} (Success: {item.get('success')}) ---")
    if item.get('success'):
        d = item.get('data', {})
        print(f"Invoice Number: {d.get('invoice_number')}")
        print(f"Vendor Name:    {d.get('vendor_name')}")
        print(f"Customer Name:  {d.get('customer_name')}")
        print(f"Invoice Total:  {d.get('invoice_amount')}")
        print(f"Customer PAN:   {d.get('customer_pan')}")
    else:
        print(f"Error: {item.get('error')}")
```

```python
import os, shutil, stat

target_dir = "/content"

# Helper to force delete read-only files if any
def remove_readonly(func, path, excinfo):
    os.chmod(path, stat.S_IWRITE)
    func(path)

print(f"Deleting all files, folders, and subfolders in {target_dir}...\n")

for item in os.listdir(target_dir):
    # Optional: Skip Colab default system folders if you want to keep them
    # If you want to delete EVERYTHING including sample_data, remove this if condition:
    if item in [".config"]:
        continue

    item_path = os.path.join(target_dir, item)
    try:
        if os.path.isdir(item_path):
            # Recursively deletes the folder and all its subfolders and files
            shutil.rmtree(item_path, onerror=remove_readonly)
            print(f"Deleted folder & subfolders: {item}/")
        else:
            # Deletes individual file or symlink
            os.remove(item_path)
            print(f"Deleted file: {item}")
    except Exception as e:
        print(f"Error deleting {item}: {e}")

print("\nFinished! Remaining in /content:")
print(os.listdir(target_dir))
```
